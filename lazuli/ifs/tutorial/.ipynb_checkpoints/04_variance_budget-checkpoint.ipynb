# 04 — The variance budget: where does the noise come from?

The variance in a simulated spectrum is not a single number — it is the sum of several
independent contributions: photon noise from the source and the sky background,
detector read-out noise, dark current, thermal background, and more. `slicersim` lets
you turn any of these off individually, which is both a useful diagnostic and literally
how the total variance budget is assembled internally.

This is notebook **4 of 5**.

```python
import numpy as np
import matplotlib.pyplot as plt
import slicersim
```

## 1. The list of variance sources

The core `Simulation` object (accessible as `target.simulation`) lists every
contribution it knows how to model.

```python
slicersim.Simulation.VARIANCE_SOURCES
```

## 2. Switching off a single source

Let's configure a SN Ia and generate its spectrum normally, then again with the
detector dark current switched off, using `switch_off=...` in `get_spectrum()`.

```python
sn = slicersim.LazuliSupernova(redshift=1.2)
sn.change_detector(nmd=(64, 8, 0), nramps=1)
print(f"exposure time: {sn.get_exposure_time():.1f} s")
```

```python
lbda, flux, variance = sn.get_spectrum()
lbda, flux_nodark, variance_nodark = sn.get_spectrum(switch_off="dark")

fig, (ax, ax_var) = plt.subplots(nrows=2, figsize=(9, 6), sharex=True)

ax.plot(lbda, flux, lw=1, label="with dark current")
ax.plot(lbda, flux_nodark, lw=1, label="dark current switched off")
ax.set_ylabel("flux [ADU]")
ax.legend()

ax_var.plot(lbda, variance, label="with dark current")
ax_var.plot(lbda, variance_nodark, label="dark current switched off")
ax_var.set_xlabel("wavelength [Å]")
ax_var.set_ylabel("variance [ADU²]")
ax_var.legend()
fig.tight_layout()
```

The gap between the two variance curves *is* the dark current's contribution to the
noise budget at each wavelength.

## 3. Doing this for every source at once

Repeating the switch-off exercise for every source, one at a time, and comparing the
resulting variances is exactly what `get_variance_contribution()` automates.

```python
%%time
import warnings
with warnings.catch_warnings():
    warnings.simplefilter("ignore")
    variance_df = sn.get_variance_contribution()
variance_df.head()
```

Each column (besides `lbda`, `flux` and `variance`) is one source's own contribution
to the variance, in the same units. Summing them recovers the total:

```python
sources = list(slicersim.Simulation.VARIANCE_SOURCES)
np.allclose(variance_df[sources].sum(axis=1), variance_df["variance"])
```

## 4. Visualizing the full budget

`simulation.show_variance_sources()` turns that table into a stacked view: the spectrum
itself, the per-wavelength SNR built up source by source, and each source's fractional
share of the total variance.

```python
fig = sn.simulation.show_variance_sources(variance_df)
```

For this SN Ia at z=1.2, the detector-related terms (`dark`, `ron`, `roic_glow`)
dominate the noise budget at essentially all wavelengths — a very different picture than
a ground-based, sky-background-dominated spectrograph.

## Recap

- `Simulation.VARIANCE_SOURCES` lists every noise contribution `slicersim` models.
- `target.get_spectrum(switch_off="dark")` (or any other source name) removes that one
  contribution, showing directly how much it matters.
- `target.get_variance_contribution()` does this systematically for every source and
  returns a table where the columns sum exactly to the total variance — this *is* how
  the variance budget is built.
- `target.simulation.show_variance_sources(df)` plots the whole budget at once.
