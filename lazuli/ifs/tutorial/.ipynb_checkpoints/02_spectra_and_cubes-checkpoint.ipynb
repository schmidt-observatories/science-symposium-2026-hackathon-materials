# 02 — Spectra, cubes, and the instrument configuration

Every `LazuliTarget`-like object gives you access to two representations of the same
simulated observation: the extracted 1D **spectrum**, and the full 3D **cube**
(wavelength x spatial x spatial) it was extracted from. This notebook shows both, and
how to inspect the instrument configuration behind them — the quantum efficiency curve
and the total system throughput.

This is notebook **2 of 5**.

```python
import numpy as np
import matplotlib.pyplot as plt
import slicersim
```

## 1. Set up a target

Same recipe as notebook 1: build a SN Ia and configure the detector to reach a target
SNR.

```python
sn = slicersim.LazuliSupernova(redshift=0.8, x1=0.0, c=0.0)
_ = sn.setup_to_snr(15, per_resolution=True, lbda_range=[4_000, 7_000], frame="rest")
print(f"exposure time: {sn.get_exposure_time():.1f} s")
```

## 2. Get the spectrum

`get_spectrum()` returns wavelength, flux and variance arrays. `unit` controls the
physical units (`"adu"`, `"flambda"`, `"fphoton"`, `"rate"`, `"framerate"`), and
`incl_error` controls whether the flux already has realistic noise added (`True`,
what you would really measure) or is the noise-free "truth" (`False`, useful to compare
against).

```python
lbda, flux, variance = sn.get_spectrum(unit="flambda", incl_error=True)
lbda, flux_true, _ = sn.get_spectrum(unit="flambda", incl_error=False)

fig, ax = plt.subplots(figsize=(9, 4))
ax.plot(lbda, flux, lw=1, color="0.6", label="realistic (noisy)")
ax.plot(lbda, flux_true, lw=1.5, color="crimson", label="true flux")
ax.fill_between(lbda, flux_true - np.sqrt(variance), flux_true + np.sqrt(variance),
                 color="crimson", alpha=0.2)
ax.set_xlabel("wavelength [Å]")
ax.set_ylabel("flux [erg/s/cm²/Å]")
ax.legend()
fig.tight_layout()
```

## 3. Get the cube

The spectrum above is extracted from a full spatial+spectral data cube — exactly what
the detector actually records. `get_cube()` gives you that cube directly, as
`(cube_signal, cube_variance)`, each of shape `(n_wavelength, n_y, n_x)`.

```python
cube, varcube = sn.get_cube(which="current")
print(f"cube shape (n_wavelength, n_y, n_x): {cube.shape}")
```

The `Simulation` object underneath (`target.simulation`) has a convenience plot,
`show_cube()`, showing the cube summed over space (left, the spectrum) and summed over
wavelength (right, the "white-light" image of the source on the IFU).

```python
fig = sn.simulation.show_cube()
```

## 4. Reading the instrument configuration

`target.simulation` exposes the full instrument model: `spectrograph`, `detector`,
`telescope`, etc. Two of the most useful things to inspect are the detector's quantum
efficiency (QE) curve and the total system throughput (the conversion from physical
flux to detected signal).

```python
simulation = sn.simulation

lbda_grid = simulation.spectrograph.lbda   # the instrument's native wavelength grid
qe = simulation.detector.qe                # quantum efficiency, callable in wavelength

fig, ax = plt.subplots(figsize=(8, 3.5))
ax.fill_between(lbda_grid, qe(lbda_grid), alpha=0.6)
ax.set_xlabel("wavelength [Å]")
ax.set_ylabel("quantum efficiency")
ax.set_ylim(0, 1)
ax.set_xlim(lbda_grid[0], lbda_grid[-1])
fig.tight_layout()
```

`get_effective_transmission()` returns the *total* system throughput: telescope,
spectrograph optics, and detector QE combined, in ADU (well, in the conversion factor
between an input flux in erg/s/cm²/Å and the detected signal).

```python
lbda_t, transmission = simulation.get_effective_transmission()

fig, ax = plt.subplots(figsize=(8, 3.5))
ax.fill_between(lbda_t, transmission, alpha=0.6, color="darkorange")
ax.set_xlabel("wavelength [Å]")
ax.set_ylabel("total throughput  [ADU / (erg/s/cm²/Å)]")
ax.set_ylim(0)
fig.tight_layout()
```

## Recap

- `target.get_spectrum(unit=..., incl_error=...)` gives you the 1D extracted spectrum,
  realistic or noise-free.
- `target.get_cube()` gives you the full `(cube_signal, cube_variance)` data cube behind
  that spectrum; `target.simulation.show_cube()` plots both views at once.
- `target.simulation` exposes the instrument model directly: `.detector.qe(lbda)` for
  the quantum efficiency, `.get_effective_transmission()` for the total throughput.
