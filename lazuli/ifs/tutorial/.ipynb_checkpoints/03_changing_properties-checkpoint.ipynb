# 03 — Changing target and detector properties

A common need is to explore *"what if"*: what if the supernova is slightly different
than expected, what if I read out the detector differently, what if I switch to the
other field of view? `slicersim` makes all of this a one-line call. This notebook shows
`change_properties()` (for the target) and `change_detector()` / `change_spectrograph()`
(for the instrument), and how each one shows up in the resulting spectrum.

This is notebook **3 of 5**.

```python
import numpy as np
import matplotlib.pyplot as plt
import slicersim
```

## 1. Changing target properties

Let's set up a SN Ia with some assumed parameters, configure the detector to reach
`snr=20`, and then pretend the *true* supernova parameters are slightly different — as
would happen in reality, since you trigger an exposure on a prediction, not on the
truth.

```python
sn = slicersim.LazuliSupernova(redshift=1.0, c=0.3, x1=0.2)
_ = sn.setup_to_snr(20, per_resolution=False, lbda_range=[6_000, 7_000], frame="rest")

lbda, flux_assumed, var_assumed = sn.get_spectrum(incl_error=False)
```

```python
# the actual target differs from what we assumed when triggering the exposure
sn.change_properties(redshift=1.05, c=0.2, x1=-2.0)

lbda, flux_true, var_true = sn.get_spectrum(incl_error=False)

fig, ax = plt.subplots(figsize=(9, 4))
ax.plot(lbda, flux_assumed, label="assumed (z=1.00, c=0.3, x1=0.2)")
ax.plot(lbda, flux_true, label="true (z=1.05, c=0.2, x1=-2.0)")
ax.set_xlabel("wavelength [Å]")
ax.set_ylabel("flux [ADU]")
ax.legend()
fig.tight_layout()
```

The detector read-out mode is unchanged (it was set for the *assumed* target), but
the spectrum now reflects the *true* properties — exactly what `change_properties()` is
for: keep the instrument configuration, change the astrophysical scene.

`change_properties()` works the same way for any target kind. For a star, the
natural thing to change is its magnitude:

```python
star = slicersim.LazuliCalSpec("gd71")

star.change_properties(mag=22, band="lssti")
lbda, flux_22, _ = star.get_spectrum(unit="flambda", incl_error=False)

star.change_properties(mag=21, band="lssti")
lbda, flux_21, _ = star.get_spectrum(unit="flambda", incl_error=False)

fig, ax = plt.subplots(figsize=(9, 4))
ax.plot(lbda, flux_22, label="mag=22 (lssti)")
ax.plot(lbda, flux_21, label="mag=21 (lssti)")
ax.set_xlabel("wavelength [Å]")
ax.set_ylabel("flux [erg/s/cm²/Å]")
ax.legend()
fig.tight_layout()
```

## 2. Changing the detector read-out mode

The detector reads out in MACC mode: `nmd = (n_groups, n_frames_per_group, n_drops)`,
repeated over `nramps` ramps. `setup_to_snr()` picks this for you, but you can set it
by hand with `change_detector()` — for instance, to see how the noise changes with a
shorter or longer integration.

```python
sn2 = slicersim.LazuliSupernova(redshift=1.2)

configs = [((10, 4, 0), 1), ((64, 8, 0), 1), ((64, 8, 0), 6)]
fig, ax = plt.subplots(figsize=(9, 4))

for nmd, nramps in configs:
    sn2.change_detector(nmd=nmd, nramps=nramps)
    lbda, flux, variance = sn2.get_spectrum(incl_error=False)
    snr = flux / np.sqrt(variance)
    exptime = sn2.get_exposure_time()
    ax.plot(lbda, snr, label=f"nmd={nmd}, nramps={nramps}  ({exptime:.0f} s)")

ax.set_xlabel("wavelength [Å]")
ax.set_ylabel("SNR per wavelength bin")
ax.legend(fontsize=8)
fig.tight_layout()
```

As expected, more groups and more ramps (i.e. a longer exposure) directly buy a
higher SNR.

## 3. Changing the spectrograph field mode

Lazuli's IFU can also be reconfigured between its two field modes, `"narrow"`
(finer spatial sampling, smaller field of view) and `"wide"` (coarser sampling, larger
field of view) — see `change_spectrograph()`. Notebook 5 explores this in depth; here
is the one-line way to switch:

```python
print("before:", sn2.get_spectrograph_field())
sn2.change_spectrograph("wide")
print("after: ", sn2.get_spectrograph_field())
```

## Recap

- `target.change_properties(**kwargs)` updates the astrophysical scene (redshift,
  stretch, color, magnitude, band, ...) without touching the instrument setup.
- `target.change_detector(nmd=..., nramps=...)` sets the detector read-out mode by hand;
  more groups/ramps means a longer exposure and a higher SNR.
- `target.change_spectrograph("narrow" | "wide")` switches the IFU field mode.
